### 🔀 Feature and Target Separation

The dataset is split into independent variables (features) and the dependent variable (target) to prepare for model training.


```python
X = df.drop('deposit', axis=1)
y = df['deposit']

X.shape, y.shape
```

### ⚖️ Feature Scaling

To ensure uniformity among numerical features, Min-Max normalization is applied using NumPy. This scales all values between 0 and 1.


```python
X = (X - np.min(X)) / (np.max(X) - np.min(X))
X.head()

```

### 🧪 Train-Test Split

The dataset is divided into training and testing sets to evaluate model performance.


```python
df = df.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
split = int(0.8 * len(X))

X_train = X[:split]
X_test = X[split:]

y_train = y[:split]
y_test = y[split:]
```

```python
majority_class = y_train.value_counts().idxmax()
y_pred = np.full(len(y_test), majority_class)

accuracy = np.mean(y_pred == y_test)
accuracy
```

##🤖 MACHINE LEARNING MODEL: RANDOM FOREST CLASSIFIER

After completing data preprocessing and establishing a baseline benchmark,
a Random Forest Classifier was implemented to improve prediction performance.

Random Forest is an ensemble learning algorithm that builds multiple decision
trees and combines their outputs to produce a more accurate and stable result.
This approach helps reduce overfitting and captures complex relationships
between features.

The model is well-suited for the Bank Marketing dataset because it efficiently
handles both numerical and categorical variables. The dataset was divided into
training and testing sets, and the model was trained on historical data to
predict whether a customer would subscribe to a term deposit.


```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report


rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)


rf_model.fit(X_train, y_train)


y_pred_rf = rf_model.predict(X_test)


rf_accuracy = accuracy_score(y_test, y_pred_rf)
rf_accuracy * 100
```

##MODEL EVALUATION

The model was evaluated using a confusion matrix and a classification report.
The confusion matrix shows correct and incorrect predictions, while the
classification report provides precision, recall, and F1-score values.

These metrics help assess how well the model distinguishes between customers
who subscribe to a term deposit and those who do not.


```python

confusion_matrix(y_test, y_pred_rf)
```

```python

print(classification_report(y_test, y_pred_rf))
```

##🟦 📈 Model Performance Comparison

The Random Forest model demonstrated a significant improvement over the baseline classifier. While the baseline model predicted only the majority class, the Random Forest classifier successfully learned patterns from the data, resulting in substantially higher predictive accuracy. This highlights the effectiveness of advanced ensemble learning techniques in solving real-world classification problems.

```python
print("Baseline Model Accuracy: 53.74%")
print("Random Forest Model Accuracy:", rf_accuracy * 100)
```
